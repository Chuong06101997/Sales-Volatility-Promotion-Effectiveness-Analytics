# Notebook 05 — Business Recommendations

**Business Question:** What should Supply Chain and Commercial teams actually do based on the evidence established in Notebook 03 (Hypothesis Testing) and Notebook 04 (Forecast Benchmark)?

This notebook performs no new statistical analysis, no hypothesis testing, and no forecasting. It converts evidence already established in Notebook 03 and Notebook 04 into business decisions. Notebook 03 remains the only statistical source of truth; Notebook 04 remains the only forecasting source of truth. Notebook 05 becomes the only source of truth for business recommendations.

## Section 1 — Business Context

Current demand planning relies heavily on judgement. Notebook 03 tested the demand patterns Supply Chain and Commercial teams had assumed were true, and established which are statistically supported. Notebook 04 evaluated whether a simple, transparent forecasting approach already provides useful planning accuracy, before any investment in more advanced forecasting. Notebook 05 converts both into concrete actions -- and is equally explicit about what the evidence does not yet support.

## Section 4 — Recommendation 1: Focus S&OP Attention on the Highest-Risk SKU-Store Combinations

**Evidence gap, stated plainly before any recommendation:** a SKU-store-level risk ranking (combining volatility and stock-out incidence) is **not present in the current, canonical Notebook 03 or Notebook 04**. Notebook 03's Section 8 (H6) evaluates volatility only at the whole-store level (daily total sales, CoV 0.1097 vs. 0.1147), and Notebook 04's diagnostics (Section 5) show forecast error by store and category, not by individual SKU-store combination. Neither is a substitute for a genuine SKU-store risk score.

**Why this matters:** recommending "focus on the highest-risk SKU-store combinations" without that ranking existing in the canonical evidence base would be recommending something stronger than the current evidence supports, which this notebook is explicitly instructed not to do.

**What the evidence does support instead:** Notebook 03 (H6) shows the two stores do not differ dramatically in volatility, and Notebook 04 shows store and category do not meaningfully affect forecast difficulty. This means a broad, store-level or category-level prioritization is **not** well supported by the current evidence -- risk, if it exists, is more likely concentrated at a finer grain (specific SKU-store combinations) than either notebook currently measures.

**Recommended next step, not a recommendation to act on yet:** compute a SKU-store-level composite risk score (combining a volatility measure and stock-out incidence, at the SKU-store grain) as an addition to Notebook 03, before this recommendation can be made with genuine evidentiary support. Until that analysis exists in the canonical notebooks, this recommendation is marked **Insufficient Data** rather than issued.

## Section 5 — Recommendation 2: Promotion Strategy

**Promotion lift:** Notebook 03 (H1) found promotion days show a $151.53 mean sales increase over non-promotion days (Mann-Whitney p<0.001) -- a real, measurable association, not proof of causation.

**No post-promotion dip:** Notebook 03 (H2) found no evidence of a sales dip in the 3 days following a promotion ($1.50 difference from baseline, p=0.4255) -- the data does not show promotions simply borrowing demand from the days immediately after.

**Recommendation: a controlled rollout, not a broad expansion.** Test a modest increase in promotion frequency on a limited set of products, with margin tracked alongside sales lift throughout the pilot. The lift finding is association, not causation -- competitor activity, in-store execution, or other unmeasured factors could be contributing, and this has not been ruled out.

**Margin monitoring is not optional.** Neither Notebook 03 nor Notebook 04 evaluated promotion cost or margin impact -- only sales lift. A pilot without margin tracking would risk mistaking a top-line lift for a profitable one.

**Explicitly avoided here:** any ROI figure for expanded promotions. No cost or margin data was evaluated in Notebooks 03-04, so none is estimated here.

## Section 6 — Recommendation 3: Forecasting

**Should planners use the Seasonal Historical Average baseline?** Yes, for routine, non-promotional planning. Notebook 04 established 28.63% WAPE overall and 27.31% WAPE specifically on non-promotion days -- a reasonable, transparent, spreadsheet-verifiable starting point.

**Should it replace planners?** Not yet. Notebook 04 explicitly states that whether this benchmark beats current planner judgement is **Insufficient Data** -- no historical record of planner forecast accuracy exists to compare against. Replacing judgement without that comparison would be a decision made without the evidence to support it.

**Should ML be adopted immediately?** No. Notebook 04, Section 8 found the benchmark's error is concentrated almost entirely in one explainable segment -- promotion days (39.25% WAPE, +$182.63 bias) -- rather than spread unexplainably across stores, categories, or holidays. A targeted fix to the existing simple benchmark is the evidence-supported next step, not a wholesale move to machine learning.

## Section 8 — Business Impact

**Planning focus:** calendar-aware buffer planning (weekend, Summer, Spring) can be adopted with high confidence, since these are the most statistically robust findings in the project.

**Promotion decision quality:** decisions can shift from assumption ("promotions probably work") to evidence ("promotions show a $151.53 mean lift, with no detected pull-forward dip in the 3 days after") -- a qualitative improvement in the basis for those decisions, not a quantified financial one.

**Forecast transparency:** the Seasonal Historical Average gives planners a benchmark they can verify by hand, replacing an unverifiable, judgement-only starting point for routine planning.

**Operational efficiency:** the clearest, most fixable gap (promotion-day forecasting) is now identified and isolated, rather than being an unknown, unlabeled source of planning error.

No ROI figure, cost saving, or revenue impact is estimated here -- Notebooks 03 and 04 did not evaluate cost data, and inventing one would exceed what either notebook established.

## Section 11 — Limitations

- **Two stores only.** Every store-related finding (Notebook 03, H6) describes these two specific stores. It cannot be generalized to other locations, formats, or markets, and no recommendation here assumes it can.
- **No customer-level data.** Nothing in this project speaks to individual customer behavior, loyalty, or repeat purchase -- recommendations are based on aggregate SKU-store-day patterns only.
- **No cost data.** Neither margin structure nor holding/inventory cost was available to Notebooks 03-04. Every recommendation involving margin (Section 5) calls for tracking it going forward, not estimating it now.
- **No causal experiment.** The promotion lift finding (Notebook 03, H1) is an association. No controlled experiment was possible, so no recommendation here treats the lift as proven cause and effect.
- **Forecast benchmark only, not a production system.** Notebook 04 evaluated a simple, transparent baseline -- not a deployed, monitored, or maintained forecasting system. Recommendation 3 (Section 6) reflects that distinction directly.

Each of these limitations directly shaped which recommendations appear in Sections 4-6 and which are explicitly withheld in Section 7 -- they are not an afterthought.

## Section 12 — Final Conclusion

**What should be done now:** adopt calendar-aware buffer planning around weekends, Summer, and Spring (the strongest evidence in this project); begin using the Seasonal Historical Average as a starting point for routine, non-promotional planning; and start logging planner forecasts prospectively so the planner-vs-benchmark comparison can finally be made.

**What should be tested next:** a controlled, margin-tracked pilot of increased promotion frequency on a limited product set; and a SKU-store-level risk score, which does not yet exist in the canonical evidence base but is a well-defined, achievable next analysis.

**What should wait:** any broad promotion expansion, any inventory increase applied everywhere, and any decision to replace planner judgement outright -- none of these have the evidence behind them yet.

**What should definitely not be done yet:** invest in machine learning forecasting or purchase commercial forecasting software. The current benchmark's error is concentrated in one identified, fixable gap, not a diffuse failure that would justify that cost.

The purpose of this project was not to maximize statistical sophistication, but to improve business decisions with evidence proportional to what the data could genuinely support.
